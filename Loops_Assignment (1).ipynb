## Student Marks Entry

### Goal:
    Record student marks and skip invalid entries.

### Steps:

    Create an empty list marks = [].
    
    Ask user to enter marks for 5 students.
    
    If marks < 0 or > 100 → print “Invalid marks” and continue.
    
    If user enters -1 → stop taking input (break).
    
    Print valid marks and their average.

### Example:

    Enter marks: 90  
    Enter marks: 105 → Invalid marks  
    Enter marks: 80  
    Enter marks: -1 → stop  
    Valid Marks: [90, 80]  
    Average = 85.0

```python
print(f"""______________________________________________________________________________________________
_______________________________________STUDENT MARKS ENTRY_______________________________________________
__________________________________________________________________________________________________________
""")
marks=[]
for i in range(5):
    students_mark = int(input('Enter student marks: '))
    if students_mark == -1:
        print('Stop Taking Input')
        break
    elif students_mark < 0 or students_mark > 100:
        print(f'Invalid Marks: {students_mark}')
        continue
    marks.append(students_mark)
print(f'Valid Marks: {marks}')
if len(marks)>0:
    average = sum(marks)/len(marks)
    print(f'Average Marks: {average}')
else:
    print('No valid marks entered')
```

## Create a menu-driven banking system using loops:
* Create Account
* Deposit
* Withdraw
* Check Balance
* Exit

```python
print(f"""_________________________________BANKING SYSTEM______________________________________
_____________________________________________________________________


                           Press 1. You Create Account
                           
                           Press 2. You Can Deposit
                           
                           Press 3. You Can Withdraw
                           
                           Press 4. Check Your Balance
                           
                           Press 5. 'Exit'
""")
balance = 0
account_created = []

while True:
    choice = int(input('enter the choice'))
    if choice == 1:
        if account_created:
            print('Account already exists')
        else:
            account_created = True
            print('Account created successfully')
    elif choice == 2:
        if not account_created:
            print('Please create an account first')
            continue
        amount = float(input('Enter deposit amount'))
        if amount > 0:
            balance += amount
            print('Deposit successful')
        else:
            print('Invalid amount')
    elif choice == 3:
        if not account_created:
            print('Please create an account first')
            continue
        amount = float(input('Enter withdrawal amount'))
        if amount <= 0:
            print('Invalid amount')
        elif amount > balance:
            print('Insufficient balance')
        else:
            balance -= amount
            print('Withdrawal successful')
    elif choice == 4:
        if account_created:
            print(f'Balance: {balance}')
        else:
            print('Please create an account first')
    elif choice == 5:
        print('Thank you for using our banking system')
        break
    else:
        print('Invalid choice')
```

```python
print(f"""______________________________Student Result Sheet___________________________
____________________________________________________________""")

for student in range(1, 6):
    marks = []
    failed_subjects = []
    print(f'\nStudent {student}')
    for subject in range(1, 5):
        mark = int(input(f'Subject {subject} marks'))
        while mark < 0 or mark > 100:
            print('Invalid marks. Enter 0-100')
            mark = int(input(f'Subject {subject} marks'))
        marks.append(mark)
        if mark < 33:
            failed_subjects.append(subject)
    print(f'Marks: {marks}')
    if failed_subjects:
        for subject in failed_subjects:
            print(f'Failed in Subject: {subject}')
        print('Result: Fail')
    else:
        if mark >= 80:
            grade = 'A'
        elif mark >= 70:
            grade = 'B'
        elif mark >= 60:
            grade = 'C'
        elif mark >= 50:
            grade = 'D'
        else:
            grade = 'E'
        print(f'total: {sum(marks)}')
        print(f'Grade: {grade}')
        print('Result: Pass')
```

## Holiday Package Evaluator
### Goal:

Compare multiple travel packages across features like hotel, transport, food.

### Steps:

Outer loop → 3 packages.

Inner loop → 3 features.

Input rating out of 5.

Skip if rating = 0.

Print average rating per package.

### Example:
Package 1: [4, 5, 5] → Avg = 4.67  
Package 2: [3, 4, 0] → Avg = 3.5

```python
print(f"""____________________________Holiday Package Evaluator____________________________
____________________________________________________________________""")

for package in range(1, 4):
    ratings = []
    print(f'\nPackage {package}')
    for feature in range(1, 4):
        rating = int(input(f'Feature {feature} rating (0-5)'))
        if rating < 0 or rating > 5:
            print('Invalid rating. Skipped')
            continue
        if rating == 0:
            continue
        ratings.append(rating)
    if len(ratings) > 0:
        average = sum(ratings) / len(ratings)
        print(f'Ratings: {ratings}')
        print(f'Average Rating: {round(average, 2)}')
    else:
        print('No valid ratings')
```
